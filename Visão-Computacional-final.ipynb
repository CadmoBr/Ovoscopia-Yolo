# Detecção e Classificação de Ovos Férteis e Inférteis Utilizando Visão Computacional

## Problema de Visão Computacional

### Descrição do Problema

O problema abordado neste projeto consiste na **detecção e classificação de ovos férteis e inférteis a partir de imagens**, utilizando técnicas de visão computacional e aprendizado profundo.

A identificação de ovos férteis é uma tarefa importante em processos de incubação e produção avícola, pois permite separar ovos que possuem embrião em desenvolvimento daqueles que não apresentam fertilização. Tradicionalmente, essa análise pode ser realizada por inspeção manual através do processo conhecido como **candling**, no qual uma fonte de luz é utilizada para observar o interior do ovo.

Neste projeto, propõe-se uma abordagem automatizada baseada em **detecção de objetos utilizando redes neurais convolucionais**, permitindo que o sistema identifique automaticamente o tipo de ovo presente na imagem.

O modelo recebe imagens contendo ovos e realiza duas tarefas principais:

- Detectar o objeto presente na imagem (ovo)
- Classificar o objeto em uma das duas categorias:

| Classe | Descrição |
|------|------|
| Fertile | Ovo fértil |
| Infertile | Ovo infértil |

Para resolver esse problema foi utilizado o modelo **YOLOv8 (You Only Look Once versão 8)**, amplamente utilizado em tarefas de detecção de objetos em tempo real.

---

# Fluxo Completo da Solução Proposta

O sistema desenvolvido segue um pipeline composto por diversas etapas, desde a obtenção do conjunto de dados até a classificação final das imagens.

## Pipeline do Sistema
Dataset de imagens
↓
Organização do dataset
(train / validation / test)
↓
Treinamento do modelo YOLOv8
↓
Avaliação do modelo
(Precision, Recall e mAP)
↓
Experimentos com hiperparâmetros
↓
Teste do modelo em imagens
↓
Classificação final
(fertile ou infertile)

## Conjunto de Dados Utilizado

O dataset utilizado foi **Dataset of Fertile and Infertile Chicken Eggs**, disponível na plataforma Kaggle.

🔗 **Link do dataset:**  
https://www.kaggle.com/datasets/mwahyuadin/dataset-of-fertile-and-infertile-chicken-eggs/data

O projeto utiliza um conjunto de imagens contendo ovos classificados em duas categorias: **férteis** e **inférteis**. Cada imagem possui anotações indicando a localização do objeto e a respectiva classe associada.

O dataset foi organizado seguindo a estrutura exigida pelo framework **YOLO**, contendo diretórios separados para imagens e rótulos.

### Divisão do Dataset

A divisão dos dados foi realizada da seguinte forma:

| Conjunto | Quantidade de Imagens | Percentual |
|--------|-----------------------|------------|
| Train | 140 imagens | 60% |
| Validation | 40 imagens | 20% |
| Test | 20 imagens | 20% |
| **Total** | **200 imagens** | **100%** |

### Função de Cada Conjunto

| Conjunto | Função |
|------|------|
| Treinamento | utilizado para treinar o modelo |
| Validação | utilizado para monitorar o desempenho durante o treinamento |
| Teste | utilizado para avaliar o modelo em dados não vistos |

## Treinamento do Modelo

Em seguida, o treinamento foi realizado utilizando o dataset configurado no arquivo **`data.yaml`**.

results = model.train(

## Predição em Imagens

O modelo treinado também foi utilizado para realizar predições diretamente em imagens.
```Python
results = model.predict(
    source="test/images",
    conf=0.25,

```
Nesse processo, o modelo analisa cada imagem e retorna:

- a localização do objeto detectado (bounding box)

- a classe identificada

- o nível de confiança da detecção

Os resultados são automaticamente salvos em uma pasta contendo as imagens com as detecções destacadas.

## Conclusão

Este trabalho apresentou uma solução baseada em **visão computacional e aprendizado profundo** para a detecção e classificação de ovos férteis e inférteis a partir de imagens.

Para isso, foi utilizado o modelo **YOLOv8**, amplamente reconhecido por sua eficiência em tarefas de detecção de objetos em tempo real. O modelo foi treinado utilizando um conjunto de imagens previamente anotadas contendo duas classes: **fertile** e **infertile**. O dataset foi dividido em conjuntos de **treinamento, validação e teste**, permitindo avaliar adequadamente o desempenho do modelo em diferentes etapas do processo.

A avaliação do modelo foi realizada utilizando métricas comuns em tarefas de detecção de objetos, como **Precision, Recall e Mean Average Precision (mAP)**. Os resultados obtidos demonstraram um desempenho elevado, com valores próximos de **1.0** para as principais métricas, indicando que o sistema foi capaz de identificar corretamente os ovos presentes nas imagens do conjunto de teste.

Além disso, foram realizados experimentos variando alguns **hiperparâmetros de treinamento**, como o número de épocas e o tamanho do batch. Os resultados mostraram que essas variações não produziram diferenças significativas no desempenho do modelo, sugerindo que o modelo já estava bem ajustado para o dataset utilizado.

Como etapa adicional de avaliação, o modelo também foi testado em **imagens extras que não faziam parte do dataset original**. Nessas imagens, o sistema foi capaz de realizar a detecção e classificação dos ovos, porém foram observados alguns comportamentos que podem indicar **possível overfitting do modelo**. Em alguns casos, o modelo apresentou **detecções adicionais ou classificações excessivamente confiantes**, mesmo em imagens que possuíam características diferentes das presentes no conjunto de treinamento.

Esse comportamento pode estar relacionado ao **tamanho relativamente reduzido do dataset e à baixa variabilidade das imagens utilizadas no treinamento**, o que pode fazer com que o modelo aprenda padrões muito específicos do conjunto de dados original.

De forma geral, os resultados demonstram que técnicas modernas de **visão computacional baseadas em redes neurais convolucionais** podem ser aplicadas com sucesso na automatização da análise de ovos.

````python
from pathlib import Path
from ultralytics import YOLO

root = Path.cwd()
data_yaml = root / 'data.yaml'

assert data_yaml.exists(), 'Arquivo data.yaml não encontrado na raiz do projeto.'
assert (root / 'train' / 'images').exists(), 'Pasta train/images não encontrada.'
assert (root / 'valid' / 'images').exists(), 'Pasta valid/images não encontrada.'
assert (root / 'test' / 'images').exists(), 'Pasta test/images não encontrada.'

model = YOLO('yolov8n.pt')

results = model.train(
    data=str(data_yaml),
    epochs=50,
    imgsz=640,
    batch=8,
    project=str(root / 'runs' / 'detect'),
    name='train_final',
    exist_ok=True,
)

metrics = model.val(data=str(data_yaml), split='test')
print(metrics.results_dict)

for img in ['ovo.jpeg', 'ovo2.jpg', 'ovo3.jpg', 'ovo4.jpg']:
    img_path = root / img
    if img_path.exists():
        model.predict(source=str(img_path), conf=0.25, save=True, project=str(root / 'runs' / 'detect'), name='predict_final')
    else:
        print(f'Imagem não encontrada: {img_path}')
````
